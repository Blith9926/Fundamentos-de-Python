```python
# 2.1 Crea aquí la función renombrar_columnas_es(df) y la función separar_por_tipo(df, tipo_emision)
def renombrar_columnas_es(df):
    df_renombrado = df.rename(columns={
        'Country': 'pais',
        'Year': 'anio',
        'Type': 'tipo_emision',
        'Value': 'valor'
    })
    return df_renombrado


def separar_por_tipo(df, tipo_emision):
    df_filtrado = df[df['tipo_emision'] == tipo_emision]
    return df_filtrado  
```

```python
# 2.2 Crea aquí df_trabajo, df_totales_es y df_per_capita_es
df_trabajo = renombrar_columnas_es(df_raw)
df_totales_es = separar_por_tipo(df_trabajo, 'Emisiones (kilotoneladas metricas de dioxido de carbono)')
df_per_capita_es = separar_por_tipo(df_trabajo, 'Emisiones per capita (toneladas metricas de dioxido de carbono)')
```

```python
# 2.3 Muestra aquí evidencia de que la preparación fue correcta
print(df_trabajo.columns)
print(df_trabajo.head(5))
print(df_totales_es.shape)
print(df_per_capita_es.shape)

```

```python
# 3.1 Escribe aquí tu prediccion_inicial y luego crea registros_totales a partir de df_totales_es
prediccion_inicial = "Creo que Taiwan tendrá el mayor valor registrado por que es un pais que usa muchos recursos y esta sobre poblado."
registros_totales = df_totales_es.to_dict(orient='records')
```

```python
# 3.2 Crea aquí la función resumen_pais(registros, pais)
def resumen_pais(registros, pais):
    registros_pais = [registro for registro in registros if registro['pais'] == pais]
    cantidad_registros = len(registros_pais)
    anio_min = min(registro['anio'] for registro in registros_pais)
    anio_max = max(registro['anio'] for registro in registros_pais)
    valor_max = max(registro['valor'] for registro in registros_pais)
    
    resumen = {
        'pais': pais,
        'cantidad_registros': cantidad_registros,
        'anio_min': anio_min,
        'anio_max': anio_max,
        'valor_max': valor_max
    }
    
    return resumen  

```

```python
# 3.3 Resuelve aquí las tareas solicitadas con Python base

paises_ordenados = sorted(set(registro['pais'] for registro in registros_totales))
valor_maximo_total = max(registro['valor'] for registro in registros_totales)
registro_valor_maximo_total = max(registros_totales, key=lambda registro: registro['valor'])
pais_valor_maximo_total = registro_valor_maximo_total['pais']
anio_max_costa_rica = max(registro['anio'] for registro in registros_totales if registro['pais'] == 'Costa Rica')

print(paises_ordenados)
print(valor_maximo_total)
print(pais_valor_maximo_total)
print(anio_max_costa_rica)
```

```python
# 3.4 Construye aquí reporte_centroamerica_base

reporte_centroamerica_base = []
paises_centroamerica = ['Costa Rica', 'Guatemala', 'Honduras', 'El Salvador', 'Nicaragua']
for anio in sorted(set(registro['anio'] for registro in registros_totales)):
    total_por_anio = sum(registro['valor'] for registro in registros_totales if registro['anio'] == anio and registro['pais'] in paises_centroamerica)
    reporte_centroamerica_base.append({'anio': anio, 'total_por_anio': total_por_anio})

print(reporte_centroamerica_base)

```

```python
# 4.1 Responde aquí las preguntas analíticas con pandas
paises_unicos = df_totales_es['pais'].nunique()
valor_maximo_global = df_totales_es['valor'].max()
pais_valor_maximo_global = df_totales_es.loc[df_totales_es['valor'].idxmax(), 'pais']  



print(paises_unicos)
print(valor_maximo_global)
print(pais_valor_maximo_global)
```

```python
# 4.2 Construye aquí reporte_total_centroamerica y reporte_promedio_centroamerica
promedio_emisiones_per_capita_2005 = df_per_capita_es[df_per_capita_es['anio'] == 2005]['valor'].mean()
print(promedio_emisiones_per_capita_2005)

reporte_total_centroamerica = df_totales_es[df_totales_es['pais'].isin(['Costa Rica', 'Guatemala', 'Honduras', 'El Salvador', 'Nicaragua'])].groupby('anio')['valor'].sum().reset_index().rename(columns={'valor': 'total_por_anio'})
print(reporte_total_centroamerica)


reporte_promedio_centroamerica = df_per_capita_es[df_per_capita_es['pais'].isin(['Costa Rica', 'Guatemala', 'Honduras', 'El Salvador', 'Nicaragua'])].groupby('anio')['valor'].mean().reset_index().rename(columns={'valor': 'promedio_por_anio'})
print(reporte_promedio_centroamerica)


```

```python
# 4.3 Construye aquí conversiones_costa_rica

conversiones_costa_rica = df_totales_es[df_totales_es['pais'] == 'Costa Rica'].copy()
conversiones_costa_rica['kilogramos'] = conversiones_costa_rica['valor'] * 1000
conversiones_costa_rica['libras'] = conversiones_costa_rica['kilogramos'] * 2.20462
conversiones_costa_rica = conversiones_costa_rica[['anio', 'valor', 'kilogramos', 'libras']]
print(conversiones_costa_rica)  



```

```python
# 4.4 Crea aquí la función comparar_paises(df, paises)

def comparar_paises(df, paises):
    df_filtrado = df[df['pais'].isin(paises)]
    return df_filtrado


comparacion_2017 = comparar_paises(df_totales_es, ['Costa Rica', 'Guatemala', 'Honduras', 'El Salvador', 'Nicaragua', 'Spain'])
comparacion_2017 = comparacion_2017[comparacion_2017['anio'] == 2017]
comparacion_2017 = comparacion_2017[['pais', 'anio', 'valor']]
comparacion_2017 = comparacion_2017.sort_values('valor', ascending=False)
print(comparacion_2017)
```
